```python
from PIL import Image
import numpy as np

# 讀入 JPG
input_jpg_path = "01_background.jpg"
jpg_image = Image.open(input_jpg_path)

# 轉成 numpy array（取得所有 pixel RGB）
pixels = np.array(jpg_image)

# 確認是 RGB 模式
if jpg_image.mode != "RGB":
    pixels = jpg_image.convert("RGB")
    pixels = np.array(pixels)

# 儲存成 BMP（無壓縮）
bmp_image = Image.fromarray(pixels, "RGB")
bmp_image.save("output.bmp")

print("已經完成 BMP 儲存！")

# 再把 BMP 轉回 JPG（觀察壓縮）
bmp_loaded = Image.open("output.bmp")

# 可以設定品質（quality越高，壓縮越小）
bmp_loaded.save("output_compressed_100.jpg", format="JPEG", quality=100)

print("已經完成 BMP -> JPG 儲存！")
```
